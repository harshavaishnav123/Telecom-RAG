```python
import json
INPUT_FILE=r"/home/central-library/Desktop/nihal/cp/data/TeleQnA.txt"
OUTPUT_FILE=r"/home/central-library/Desktop/nihal/cp/data/processed_data/teleqna_processed.csv"
with open(INPUT_FILE,"r",encoding="utf-8") as f:
    data=json.load(f)
new_data={}
for key,item in data.items():
    options=[]
    i=1
    while f"option {i}" in item:
        options.append(item[f"option {i}"])
        i+=1

    answer=item["answer"]
    if ":" in answer:
        answer=answer.split(":",1)[1].strip()
    new_data[key]={
        "question": item["question"],
        "options":options,
        "answer":answer,
        "explanation":item.get("explanation",""),
        "category":item.get("explanation","")
    }
with open(OUTPUT_FILE,"w",encoding="utf-8") as f:
    json.dump(new_data,f,indent=4,ensure_ascii=False)
```
